```python
import pandas as pd

# --- BLOK 1: BACA DATA & PILIH FITUR DASAR ---
def load_and_select(path):
    dataset = pd.read_csv(path)
    print("Shape awal:", dataset.shape)
    print("Kolom:", dataset.columns.tolist())

    # hanya ambil fitur & target yang akan dipakai
    data = dataset[["age", "gender", "at_risk"]].copy()
    print("Shape setelah seleksi kolom:", data.shape)
    return data

data_selected = load_and_select("stroke_risk_dataset_awal.csv")
```

```python
# --- BLOK 2: PEMBERSIHAN UMUM (missing value) ---
def clean_missing(data):
    before = data.shape[0]
    data = data.dropna()
    after = data.shape[0]
    print(f"Drop NA: {before} -> {after} baris")
    return data

data_selected = clean_missing(data_selected)
```

```python
# --- BLOK 3: PENGOLAHAN TARGET Risiko Stroke ---
def make_stroke_label(data):
    stroke_mapping = {
        0: "Normal",
        1: "Risiko Stroke"
    }
    data["at_risk"] = data["at_risk"].astype(int)
    data["StrokeStatus"] = data["at_risk"].map(stroke_mapping)
    print("Unique StrokeStatus:", data["StrokeStatus"].unique())
    return data

data_selected = make_stroke_label(data_selected)
```

```python
# --- BLOK 4: PENGOLAHAN FITUR JenisKelamin ---
def make_gender_label(data):
    sex_mapping = {
        "Female": "Perempuan",
        "Male": "Laki-Laki"
    }
    data["JenisKelamin"] = data["gender"].map(sex_mapping)
    print("Unique JenisKelamin:", data["JenisKelamin"].unique())
    return data

data_selected = make_gender_label(data_selected)
```

```python
# --- BLOK 5: PENGOLAHAN FITUR Usia (kelompok) ---
def make_age_group(data):
    age_bins = [0, 17, 24, 29, 34, 39, 44, 49, 54, 59, 64, 69, 74, 79, 200]
    age_labels = [
        "0-17","18-24","25-29","30-34","35-39","40-44","45-49",
        "50-54","55-59","60-64","65-69","70-74","75-79","80+"
    ]
    data["Usia"] = pd.cut(
        data["age"],
        bins=age_bins,
        labels=age_labels,
        right=True
    )
    print("Unique Usia:", data["Usia"].unique())
    return data

data_selected = make_age_group(data_selected)
```

```python
# --- BLOK 6: SUSUN DATA AKHIR & SIMPAN CSV ---
def build_final_and_save(data, out_path="stroke.csv"):
    data_final = data[["StrokeStatus", "JenisKelamin", "Usia"]].copy()
    print("Contoh data_final:")
    display(data_final.head(15))

    data_final.to_csv(out_path, index=False)
    print(f"File '{out_path}' berhasil disimpan.")
    return data_final

data_final = build_final_and_save(data_selected, "stroke.csv")
```
